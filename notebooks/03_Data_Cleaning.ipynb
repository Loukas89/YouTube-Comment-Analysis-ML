```python
#Πρώτα ένα check του file:

import pandas as pd
from IPython.display import display

data = pd.read_csv(r"C:\Users\Owner\Desktop\Essex\03. Main\Data\Python\Group Project\Data Clean\youtube_comments_annotated_cleaned.csv")

print(data.info())
print('-'*90)
print(data.columns)
print('-'*90)
print(data.head())

#Έχουμε 0 nulls και οι στύλες που θέλουμε είναι εκεί.
```

```python
#Μετά ένα γρήγηρο check nulls & Duplicates on commend_id only γιατί πρέπει να είναι unique.

print(data.isnull().sum())
print('-'*90)
print("Duplicate comment IDs:", data['comment_id'].duplicated().sum())

#Όλα οκ.
```

```python
#Closer look at the file:

print("Topic counts:")
print(data['topic'].value_counts())
print("\nSentiment counts:")
print(data['sentiment_label'].value_counts())
data[['comment_id', 'topic', 'text', 'sentiment_label']].head(10)

#sentiment_label έχει όντως 1000 Values και κανένα Nan ενώ είναι μόνο 3 entries: Positive/Negative/Neutral
```

```python
#New column to the file named as "clean_text". Αυτό για να μπορέσουμε στο τέλος να συγκρίνουμε.

data['clean_text'] = data['text']  #<---Νέο column ίδιο με το text.
data['clean_text'] = data['clean_text'].str.lower()  #<----Lowering strings texts
data[['text', 'clean_text']].head(10)  #<<---print 10 lines for check

#OK
```

```python
#URL removal with regex

data['clean_text'] = data['clean_text'].str.replace(r'http\S+|www\.\S+','',regex=True)
display(data[['text', 'clean_text']].head(10))
```

```python
#Από τις 10 πρώτες rows δεν έχει urls, ας τα βρούμε.

url_rows = data[data['text'].str.contains(r'http\S+|www\.\S+',regex=True,na=False)]
display(url_rows[['text', 'clean_text']].head(10))

#URLs removed confirmed.
```

```python
#Removing "mentions".Find & Remove to make sure.

data['clean_text'] = data['clean_text'].str.replace(r'@\w+','',regex=True)

mention_rows = data[data['text'].str.contains(r'@\w+', regex=True, na=False)]
display(mention_rows[['text', 'clean_text']].head(10))
print('-'*80)
print("Rows with mentions:", len(mention_rows))
print("Mentions remaining:",data['clean_text'].str.contains(r'@\w+', regex=True, na=False).sum())
```

```python
#Remove hashtags , same way.

data['clean_text'] = data['clean_text'].str.replace(r'#\w+','',regex=True)

hashtag_rows = data[data['text'].str.contains(r'#\w+', regex=True, na=False)]
display(hashtag_rows[['text', 'clean_text']].head(10))
print('-'*80)
print("Rows with hashtags:", len(hashtag_rows))
print("Hashtags remaining:",data['clean_text'].str.contains(r'#\w+', regex=True, na=False).sum())
```

```python
#Remove special characters. (! ? . , : ; ( ) [ ] " ' / \ ❤️ 😂 🔥)
#Από τη στιγμή που τα commends είναι annotated εγώ δεν βλέπω λόγο να κρατήσουμε emojis. Οπότε και φύγανε.

data['clean_text'] = data['clean_text'].str.replace(r'[^\w\s]|_',' ',regex=True)
special_rows = data[data['text'].str.contains(r'[^\w\s]|_', regex=True, na=False)]
display(special_rows[['text', 'clean_text']].head(10))
print('-'*80)
print("Rows with special characters:", len(special_rows))
print("Special characters remaining:",data['clean_text'].str.contains(r'[^\w\s]|_', regex=True, na=False).sum())
```

```python
#Import NLTK for "stopwords"

import nltk
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
```

```python
#Cleaning with NLTK stopwords

data['clean_text'] = data['clean_text'].apply(lambda text: ' '.join(word for word in text.split()
                                                                    if word not in stop_words))
display(data[['text', 'clean_text']].head(10))
```

```python
#Removing whitespace (multiple spaces in the text)

data['clean_text'] = data['clean_text'].str.replace(r'\s+',' ',regex=True).str.strip()
display(data[['text', 'clean_text']].head(10))
print('-'*80)
print("Rows with multiple spaces remaining:",data['clean_text'].str.contains(r'\s{2,}', regex=True, na=False).sum())
```

```python
#Inspecting for empty entries after the whole pipeline of cleaning.

empty_rows = data[data['clean_text'].str.strip() == '']
print("Empty clean_text rows:", len(empty_rows))
print('-'*80)
display(empty_rows[['text', 'clean_text', 'sentiment_label']].head(10))
print('-'*80)
display(empty_rows[['text', 'topic', 'sentiment_label']].head(20))
```

```python
#Quick check of the dataset after cleaning text column to clean_text column.

print("Final shape:", data.shape)
print('-' * 80)
print("Missing values in key columns:")
print(data[['comment_id', 'text', 'clean_text', 'topic', 'sentiment_label']].isnull().sum())
print('-' * 80)
print("Duplicate comment IDs:", data['comment_id'].duplicated().sum())
print('-' * 80)
print("Empty clean_text rows:", (data['clean_text'].str.strip() == '').sum())
print('-' * 80)
print("Topic counts:")
print(data['topic'].value_counts())
print('-' * 80)
print("Sentiment counts:")
print(data['sentiment_label'].value_counts())

#File remained same and balanced.
```

```python
#Saving changes to csv file.

data.to_csv(r"C:\Users\Owner\Desktop\Essex\03. Main\Data\Python\Group Project\Data Clean\youtube_comments_annotated_cleaned.csv",index=False)
print("File saved successfully.")
```
